### Linear Discriminant Analysis

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.metrics import confusion_matrix
from sklearn.model_selection import train_test_split
```

```python
X_multi = df.loc[:,df.columns != 'Sold']
y_multi = df['Sold']
```

```python
clf_lda = LinearDiscriminantAnalysis()
clf_lda.fit(X_multi, y_multi)
```

```python
y_pred_lda = clf_lda.predict(X_multi)
```

```python
y_pred_lda
```

```python
confusion_matrix(y_multi,y_pred_lda)
#rows represent actual class and column represent predicted class
# 192 house not sold in actual data and also not sold in predicted data
#meanwhile 79 houses sold in actual data but not sold in predicted data
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_multi, y_multi, test_size=0.2, random_state=0)
```

```python
clf_lda_split = LinearDiscriminantAnalysis()
clf_lda_split.fit(X_train, y_train)
```

```python
y_test_pred_lda_split = clf_lda_split.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
confusion_matrix(y_test, y_test_pred_lda_split)
#rows represent actual class and column represent predicted class
# 34 house not sold in actual data and also not sold in predicted data
#meanwhile 11 houses sold in actual data but not sold in predicted data
```

```python
accuracy_score(y_test, y_test_pred_lda_split)
```
