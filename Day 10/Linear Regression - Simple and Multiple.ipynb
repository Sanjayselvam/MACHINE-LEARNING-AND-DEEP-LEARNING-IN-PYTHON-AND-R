```python
import statsmodels.api as sn
```

Now we need to find out x variable from our dataset. As we already seen, the correlation coefficient of room_num with price is higher. So we are taking room_num as our variable. 

```python
X = sn.add_constant(df['room_num']) #we are adding beta0 constant to our X variable room_num
```

Fiting the model

```python
lm = sn.OLS(df['price'], X).fit() #OLS - ordinary least square
```

```python
lm.summary()
```

P>|t| is nearly zero. So, we can say there is a significant relationship between room_num and price variable

```python
from sklearn.linear_model import LinearRegression #This is the most common method of LinearRegression running
```

sklearn is a free machine learning library for the Python programming lanuguage and it supports various classification, regression and clustering algorithm

```python
y = df['price'] # y is the target variable
```

```python
X = df[['room_num']] # X is the variable to determine the target variable and it needs to two dimensional array.
#That's why we used two sqaure brackets
```

```python
lm2 = LinearRegression()
```

```python
lm2.fit(X,y)
```

```python
print(lm2.intercept_, lm2.coef_)
```
