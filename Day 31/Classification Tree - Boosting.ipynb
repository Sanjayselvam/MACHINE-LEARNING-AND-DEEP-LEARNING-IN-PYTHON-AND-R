#### Missing Value Imputation

```python
df['Time_taken'].mean()
```

```python
df['Time_taken'].fillna(value = df['Time_taken'].mean(), inplace=True)
```

For permanent changes in the dataframe, we use 'inplace=True'

```python
df.info()
```

#### Dummy variable creation

```python
df.head()
```

```python
df = pd.get_dummies(df, columns=['3D_available','Genre'], drop_first=True)
#since, we want n-1 variables for n categories we are using drop_first=True 
```

```python
df.head()
```

#### X-y Split

```python
X = df.loc[:,df.columns != 'Start_Tech_Oscar']
type(X)
```

For 'loc' function, use rows and then columns. In the above [:] that means include all rows and then we specified column after comma

```python
X.head()
```

```python
X.shape
```

```python
y = df["Start_Tech_Oscar"]
```

```python
y.head()
```

```python
y.shape
```

```python
type(y)
```

#### Test-Train Split

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2, random_state=0)
```

```python
X_train.head()
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
y_train.shape
```

### Gradient Boosting

```python
from sklearn.ensemble import GradientBoostingClassifier
```

```python
gbc_clf = GradientBoostingClassifier()
gbc_clf.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score, mean_squared_error
```

```python
accuracy_score(y_test, gbc_clf.predict(X_test))
```

```python
gbc_clf2 = GradientBoostingClassifier(learning_rate=0.02, n_estimators=1000, max_depth=1)
gbc_clf2.fit(X_train, y_train)
```

```python
accuracy_score(y_test, gbc_clf2.predict(X_test))
```

### Ada Boost

```python
from sklearn.ensemble import AdaBoostClassifier
```

```python
ada_clf = AdaBoostClassifier(learning_rate=0.02, n_estimators=5000)
```

```python
ada_clf.fit(X_train,y_train)
```

```python
accuracy_score(y_test, ada_clf.predict(X_test))
```

```python
from sklearn.ensemble import RandomForestClassifier
rf_clf = RandomForestClassifier(n_estimators=1000, n_jobs=-1, random_state=42)
rf_clf.fit(X_train, y_train)
```

```python
ada_clf2 = AdaBoostClassifier(rf_clf, learning_rate=0.05, n_estimators=500)
```

```python
ada_clf2.fit(X_train,y_train)
```

```python
accuracy_score(y_test, ada_clf2.predict(X_test))
```

```python
xgb_clf = xgb.XGBClassifier(n_estimators = 250, learning_rate=0.1, random_state=42)
```

```python
param_test1 = {
    'max_depth' : range(3,10,2),
    'gamma' : [0.1,0.2,0.3],
    'subsample': [0.8,0.9],
    'colsample_bytree': [0.8,0.9],
    'reg_aplha' : [1e-2, 0.1, 1]
}

#subsample here means using 80% and 90% of random data
#gamma also similiar to minimum samples split
#colsample_bytree means use 80% and 90% of features
#reg_alpha - regularization parameter
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
grid_search = GridSearchCV(xgb_clf, param_test1, n_jobs=-1, cv=5, scoring='accuracy')
```

```python
grid_search.fit(X_train, y_train)
```

```python
grid_search.best_params_
```

```python
cxgb_clf = grid_search.best_estimator_
```

```python
accuracy_score(y_test, cxgb_clf.predict(X_test))
```
