### Logistic Regression

```python
X = df[['price']]
```

```python
y = df['Sold']
```

```python
X.head()
```

```python
y.head()
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
clf_lr = LogisticRegression()
```

```python
clf_lr.fit(X,y)
```

```python
clf_lr.coef_ #value of beta1
```

```python
clf_lr.intercept_ #value of beta0
```

Another method

```python
import statsmodels.api as sn
```

```python
X_cons = sn.add_constant(X)
```

```python
X_cons.head()
```

```python
import statsmodels.discrete.discrete_model as sm
```

```python
logit = sm.Logit(y,X_cons).fit()
```

```python
logit.summary()
```

coef of const is beta0 and coef of price is beta1

### Logistic Regression with Multiple Predictors

```python
X_multi = df.loc[:,df.columns != 'Sold']
```

```python
X_multi.head()
```

```python
y_multi = df['Sold']
```

```python
y_multi.head()
```

```python
clf_lrs_m = LogisticRegression()
```

```python
clf_lrs_m.fit(X_multi,y_multi)
```

```python
clf_lrs_m.coef_ #value of beta1, beta2,...betan
```

```python
clf_lrs_m.intercept_ #value of beta0
```

```python
import statsmodels.api as sn
import statsmodels.discrete.discrete_model as sm
```

```python
X_cons_m = sn.add_constant(X_multi)
```

```python
logit = sm.Logit(y_multi, X_cons_m).fit()
```

```python
logit.summary()
```
