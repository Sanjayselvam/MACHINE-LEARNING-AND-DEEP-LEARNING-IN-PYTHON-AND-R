```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("C:/Users/HP/Downloads/Data Science/SDT/Movie_classification.csv", header=0)
```

#### Missing Value Imputation

```python
df['Time_taken'].mean()
```

```python
df['Time_taken'].fillna(value = df['Time_taken'].mean(), inplace=True)
```

For permanent changes in the dataframe, we use 'inplace=True'

```python
df.info()
```

#### Dummy variable creation

```python
df.head()
```

```python
df = pd.get_dummies(df, columns=['3D_available','Genre'], drop_first=True)
#since, we want n-1 variables for n categories we are using drop_first=True 
```

```python
df.head()
```

#### X-y Split

```python
X = df.loc[:,df.columns != 'Start_Tech_Oscar']
type(X)
```

For 'loc' function, use rows and then columns. In the above [:] that means include all rows and then we specified column after comma

```python
X.head()
```

```python
X.shape
```

```python
y = df["Start_Tech_Oscar"]
```

```python
y.head()
```

```python
y.shape
```

```python
type(y)
```

#### Test-Train Split

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2, random_state=0)
```

```python
X_train.head()
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

#### Training Classification Tree

```python
from sklearn import tree
clftree = tree.DecisionTreeClassifier(max_depth=3)
```

```python
clftree.fit(X_train,y_train)
```

#### Predict using trained model

```python
y_train_pred = clftree.predict(X_train)
```

```python
y_test_pred = clftree.predict(X_test)
```

```python
y_test_pred
```

#### Model performance

For regression, we can use MSE and r2 to evaluate model performance.
For Classification, we can use accuracy score and confusion matrix to evaluate model performance

```python
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
confusion_matrix(y_train, y_train_pred)
```

Rows represents actual values and columns represent predicted values
First column is predicted zero category and second column is predicted one category
First row represents actual zero and second row represents actual one


172 is true negative. 126 true value is one but predicted value is zero and False negatives because we predicted as zero but actual value is one.

14 true value is zero but predicted value is one and False positives because we predicted as one  but actual value is zero. 92 is true positive.

```python
confusion_matrix(y_test, y_test_pred)
```

```python
accuracy_score(y_test, y_test_pred)
```

#### Plotting decision Tree

```python
dot_data = tree.export_graphviz(clftree, out_file=None, feature_names=X_train.columns, filled = True)
```

```python
from IPython.display import Image
```

```python
import pydotplus
```

```python
graph = pydotplus.graph_from_dot_data(dot_data)
Image(graph.create_png())
```

Blue represents high purity of 1 and orange represents high purity of 0

#### Controlling Tree Growth

```python
clftree2 = tree.DecisionTreeClassifier(min_samples_leaf= 20, max_depth=4)
clftree2.fit(X_train, y_train)
dot_data2 = tree.export_graphviz(clftree2, out_file=None, feature_names=X_train.columns, filled = True)
graph2 = pydotplus.graph_from_dot_data(dot_data2)
Image(graph2.create_png())
```

```python
accuracy_score(y_test, clftree2.predict(X_test))
```

```python
confusion_matrix(y_test, clftree2.predict(X_test))
```
